```python
#Ciclo for con intervalos, donde 2 es el intervalo y el primer 2 es la posición inicial y el 11 es la posición final

for i in range(2,11,2):
    print(f"{i} - Dracarys😮")
```

```python
mensaje =  input("Escribe tu mensaje : ")
repeticion = int(input("Cuántas veces quieres repetir el mensaje :"))

for i in range(repeticion):
    print(f"{i+1} - {mensaje}")
```

```python
#Preguntar al profe cuántas notas quiere registrar.
#Hacer el promedio de las notas y mostrarlo
#Promedio >=3.5 Mostrar estudiante Ganó - contrario Perdió

print("== Sistema de calificacion.==")
estudiante = input("Nombre del Estudiante: ")
can_notas = int(input("Cuantas notas vas a registrar :"))

promedio = 0
for i in range(can_notas):
    notas = float(input(f"Ingrese la nota {i+1} :"))
    promedio += notas

promedio_final = promedio/can_notas
if (promedio_final) >= 3.5:
    print(f"El estudiante {estudiante} - Promedio {promedio_final:.1f} Gano ")
else:
    print(f"El estudiante {estudiante} - Promedio {promedio_final:.1f} Perdio ❌")
   
```

# BREAK -> Parar ciclo forzado

```python
#Preguntar al profe cuántas notas quiere registrar.
#Hacer el promedio de las notas y mostrarlo
#Promedio >=3.5 Mostrar estudiante Ganó - contrario Perdió

print("== Sistema de calificacion.==")
estudiante = input("Nombre del Estudiante: ")
can_notas = int(input("Cuantas notas vas a registrar :"))

promedio = 0
for i in range(can_notas):
    notas = float(input(f"Ingrese la nota {i+1} :"))
    if notas not in range (0,5):
        print("Nota inválida")
        break
    promedio += notas

    promedio_final = promedio/can_notas
    if (promedio_final) >= 3.5:
        print(f"El estudiante {estudiante} - Promedio {promedio_final:.1f} Gano ")
    else:
        print(f"El estudiante {estudiante} - Promedio {promedio_final:.1f} Perdio ❌")
```

# CICLO WHILE

```python
#Ciclo infinito
while True :
    print("Hoy es viernes y el cuerpo lo sabe 💃🕺💃")
```

```python
#Ciclo While con condicional
while True:
    menu = int(input("""
    Seleccione Opcion a realizar:

    1. Sumar
    2. Restar
    3. Salir
    :     """))

    if menu == 1:
        n1 = int(input("Ingrese n1 :"))
        n2 = int(input("Ingrese n2: "))
        print(f"Resultado {n1+n2}")

    elif menu == 2 :
        n1 = int(input("Ingrese un numero :"))
        n2 = int(input("Ingrese un numero : "))
        print(f"Resultado {n1+n2}")

    elif menu == 3 :
        print("Saliendo del sistema")
        break
    else:
        print("Opcion inválida")

```
