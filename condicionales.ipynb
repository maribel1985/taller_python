```python
# Ejercicio 3: Determinar si un número es par o impar
numero = int(input("Ingrese un número entero: "))

if numero % 2 == 0:      # si el residuo es 0 → es par
    print(f"{numero} es par")
else:
    print(f"{numero} es impar")
```

```python
# 4. Elaborar un algoritmo que solicite el nombre de una ciudad y la temperatura actual en grados Celsius, y la clasifique según estos rangos.

# - Menor a 10°C ---> "Muy fría"
# - De 10°C a 17°C ---> "Fría"
# - De 18°C a 25°C ---> "Templado"
# - De 26°C a 32°C ---> "Caliente"
# - Mayor a 32°C ---> "Muy caliente"

print("Ejercicio temparatura actual en ciudades")

ciudad = input("Nombre de la ciudad")
temperatura = int(input("Temperatura en Celsius"))

if temperatura <= 10 :
    print(f"La ciudad de {ciudad} tiene una temperatura actual de {temperatura} grados celsius, está muy fría, se recomienda estar muy bien abrigado")

elif temperatura in range (10, 17) :
    print(f"La ciudad de {ciudad} tiene una temperatura actual de {temperatura} grados celsius, está fría, se recomienda usar un abrigo")

elif temperatura in range (18, 25) :
    print(f"La ciudad de {ciudad} tiene una temperatura actual de {temperatura} grados celsius, está templada, no es necesario estar abrigado")

elif temperatura in range (26, 32) :
    print(f"La ciudad de {ciudad} tiene una temperatura actual de {temperatura} grados celsius, está caliente, se recomienda usar ropa cómoda")

else :
    print(f"La ciudad de {ciudad} tiene una temperatura actual de {temperatura > 32} grados celsius, está muy caliente, se recomienda hidratarse constantemente")
    
    
    
    
```
